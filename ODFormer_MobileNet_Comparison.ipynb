```python
# ============================================================================
# KAGGLE/COLAB INSTALLATION (Copy this to cell 1, run, then restart kernel)
# ============================================================================
"""

# Then install PyTorch and other packages
!pip uninstall -y torch torchvision timm torchaudio
!pip install torch==2.0.1 torchvision==0.15.2 --index-url https://download.pytorch.org/whl/cu118
!pip install timm==0.9.16 albumentations opencv-python-headless "matplotlib<3.9" tqdm pillow scikit-learn
!pip install "numpy<2.0" --force-reinstall

# ⚠️ IMPORTANT: RESTART KERNEL after running above commands!
# Then run this script or call main()
"""
# ============================================================================

"""
ODFormer for REFUGE Dataset - MobileNet Comparison Version
===========================================================

This implementation uses ODFormer architecture with MobileNet training configuration
for direct comparison with MobileNet-UNet results.

Pre-trained Weights:
====================
- Swin-Tiny backbone: Automatically downloaded by timm library
- No manual download required!
- Pre-trained on ImageNet-1K (28M images, 1000 classes)

mmsegmentation:
===============
- NOT required for this implementation
- This is a standalone PyTorch version
- Uses timm + PyTorch instead of mmseg framework
- Easier to install and run on Kaggle/Colab

Key Configuration (Matching MobileNet):
========================================
- Input size: 256×256 (same as MobileNet)
- Split: 600 train, 200 val, 400 test (same as MobileNet, random seed=42)
- Batch size: 8 (same as MobileNet)
- Epochs: 100 (same as MobileNet)
- Learning rate: 1e-3 with ReduceLROnPlateau (same as MobileNet)
- Loss: Categorical Cross-Entropy (same as MobileNet)
- Augmentation: Same as MobileNet
- Optimizer: Adam (same as MobileNet)

Differences from Original ODFormer:
====================================
- Input size: 256×256 instead of 512×2048
- No gradient accumulation (batch_size=8 fits in memory)
- No mixed precision (for fair comparison with MobileNet)
- Split method: Random split like MobileNet (not sequential first 600)
- Uses timm Swin-Tiny instead of mmseg version
"""

import os
import cv2
import numpy as np
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from tqdm import tqdm
import timm
import albumentations as A
from albumentations.pytorch import ToTensorV2
from sklearn.model_selection import train_test_split
from sklearn.metrics import precision_score, recall_score, f1_score, confusion_matrix

# ============================================================================
# MSCA (Multi-scale Context Aggregator)
# ============================================================================

class MSCA(nn.Module):
    """Multi-scale Context Aggregator with dilated convolutions"""
    def __init__(self, channels, dilation_rates=[1, 2, 3, 5]):
        super(MSCA, self).__init__()
        self.channels = channels
        self.dilation_rates = dilation_rates
        
        # Multi-scale dilated convolutions
        self.convs = nn.ModuleList([
            nn.Sequential(
                nn.Conv2d(channels, channels // len(dilation_rates), 3, 
                         padding=rate, dilation=rate, bias=False),
                nn.BatchNorm2d(channels // len(dilation_rates)),
                nn.GELU()
            ) for rate in dilation_rates
        ])
        
        # Fusion layer
        self.fusion = nn.Sequential(
            nn.Conv2d(channels, channels, 1, bias=False),
            nn.BatchNorm2d(channels),
            nn.GELU()
        )
    
    def forward(self, x):
        # Multi-scale feature extraction
        multi_scale = [conv(x) for conv in self.convs]
        
        # Concatenate and fuse
        concat = torch.cat(multi_scale, dim=1)
        out = self.fusion(concat)
        
        # Residual connection
        return x + out

# ============================================================================
# LBFR (Lightweight Bidirectional Feature Recalibrator)
# ============================================================================

class LBFR(nn.Module):
    """Lightweight Bidirectional Feature Recalibrator"""
    def __init__(self, channels, reduction=16):
        super(LBFR, self).__init__()
        
        # Channel attention
        self.avg_pool = nn.AdaptiveAvgPool2d(1)
        self.max_pool = nn.AdaptiveMaxPool2d(1)
        self.fc = nn.Sequential(
            nn.Conv2d(channels, channels // reduction, 1, bias=False),
            nn.ReLU(),
            nn.Conv2d(channels // reduction, channels, 1, bias=False)
        )
        
        # Spatial attention
        self.spatial_conv = nn.Sequential(
            nn.Conv2d(2, 1, 7, padding=3, bias=False),
            nn.Sigmoid()
        )
        
        # Recalibration
        self.recalibrate = nn.Sequential(
            nn.Conv2d(channels, channels, 1, bias=False),
            nn.BatchNorm2d(channels)
        )
    
    def forward(self, x):
        # Channel attention
        avg_out = self.fc(self.avg_pool(x))
        max_out = self.fc(self.max_pool(x))
        channel_att = torch.sigmoid(avg_out + max_out)
        x_channel = x * channel_att
        
        # Spatial attention
        avg_pool = torch.mean(x_channel, dim=1, keepdim=True)
        max_pool, _ = torch.max(x_channel, dim=1, keepdim=True)
        spatial_att = self.spatial_conv(torch.cat([avg_pool, max_pool], dim=1))
        x_spatial = x_channel * spatial_att
        
        # Recalibrate
        out = self.recalibrate(x_spatial)
        
        # Residual connection
        return x + out

# ============================================================================
# PPM (Pyramid Pooling Module)
# ============================================================================

class PPM(nn.Module):
    """Pyramid Pooling Module for multi-scale global context"""
    def __init__(self, in_channels, out_channels, pool_scales=[1, 2, 3, 6]):
        super(PPM, self).__init__()
        self.pool_scales = pool_scales
        
        self.stages = nn.ModuleList([
            nn.Sequential(
                nn.AdaptiveAvgPool2d(scale),
                nn.Conv2d(in_channels, out_channels // len(pool_scales), 1, bias=False),
                nn.BatchNorm2d(out_channels // len(pool_scales)),
                nn.GELU()
            ) for scale in pool_scales
        ])
        
        self.fusion = nn.Sequential(
            nn.Conv2d(in_channels + out_channels, out_channels, 3, padding=1, bias=False),
            nn.BatchNorm2d(out_channels),
            nn.GELU()
        )
    
    def forward(self, x):
        h, w = x.size(2), x.size(3)
        
        # Multi-scale pooling
        ppm_outs = [x]
        for stage in self.stages:
            pooled = stage(x)
            upsampled = F.interpolate(pooled, size=(h, w), mode='bilinear', align_corners=False)
            ppm_outs.append(upsampled)
        
        # Concatenate and fuse
        out = torch.cat(ppm_outs, dim=1)
        out = self.fusion(out)
        
        return out

# ============================================================================
# UPerNet Decoder Block
# ============================================================================

class UPerNetDecoderBlock(nn.Module):
    """UPerNet decoder block with MSCA and LBFR modules"""
    def __init__(self, in_channels, skip_channels, out_channels, use_msca=True, use_lbfr=True):
        super(UPerNetDecoderBlock, self).__init__()
        self.use_msca = use_msca
        self.use_lbfr = use_lbfr
        
        # Channel alignment
        self.input_align = nn.Sequential(
            nn.Conv2d(in_channels, out_channels, 1, bias=False),
            nn.BatchNorm2d(out_channels)
        )
        
        if skip_channels > 0:
            self.skip_align = nn.Sequential(
                nn.Conv2d(skip_channels, out_channels, 1, bias=False),
                nn.BatchNorm2d(out_channels)
            )
        else:
            self.skip_align = None
        
        # MSCA module
        if use_msca:
            self.msca = MSCA(out_channels)
        
        # Refinement convolutions
        self.conv1 = nn.Sequential(
            nn.Conv2d(out_channels, out_channels, 3, padding=1, bias=False),
            nn.BatchNorm2d(out_channels),
            nn.GELU(),
            nn.Dropout2d(0.1)
        )
        
        self.conv2 = nn.Sequential(
            nn.Conv2d(out_channels, out_channels, 3, padding=1, bias=False),
            nn.BatchNorm2d(out_channels),
            nn.GELU()
        )
        
        # LBFR module
        if use_lbfr:
            self.lbfr = LBFR(out_channels)
    
    def forward(self, x, skip=None):
        # Align input channels
        x = self.input_align(x)
        
        # Upsample
        x = F.interpolate(x, scale_factor=2, mode='bilinear', align_corners=False)
        
        # Add skip connection
        if skip is not None and self.skip_align is not None:
            skip_aligned = self.skip_align(skip)
            x = x + skip_aligned
        
        # MSCA
        if self.use_msca:
            x = self.msca(x)
        
        # Refinement
        x = self.conv1(x)
        x = self.conv2(x)
        
        # LBFR
        if self.use_lbfr:
            x = self.lbfr(x)
        
        return x

# ============================================================================
# ODFormer Model (Adapted for 256×256 input)
# ============================================================================

class ODFormer(nn.Module):
    """
    ODFormer adapted for 256×256 input (MobileNet comparison)
    
    Args:
        num_classes: Number of segmentation classes
        pretrained: Use pre-trained Swin weights
        backbone: Swin Transformer variant
    """
    def __init__(self, num_classes=3, pretrained=True, backbone='swin_tiny_patch4_window7_224'):
        super(ODFormer, self).__init__()
        
        # Swin Transformer backbone from timm
        # Set strict_img_size=False to allow different input sizes
        self.backbone = timm.create_model(
            backbone,
            pretrained=pretrained,
            features_only=True,
            out_indices=(0, 1, 2, 3),
            strict_img_size=False,  # Allow flexible input sizes
            img_size=256  # Set expected size to 256
        )
        
        # Get feature dimensions - use a dummy forward pass
        with torch.no_grad():
            dummy_input = torch.randn(1, 3, 256, 256)
            dummy_feats = self.backbone(dummy_input)
            # Swin outputs in [B, H, W, C] format, need to get C
            # Check the actual shape format
            if len(dummy_feats[0].shape) == 4 and dummy_feats[0].shape[1] < dummy_feats[0].shape[-1]:
                # Format is [B, H, W, C]
                feat_dims = [f.shape[-1] for f in dummy_feats]
            else:
                # Format is [B, C, H, W]
                feat_dims = [f.shape[1] for f in dummy_feats]
        
        print(f"  Detected feature dimensions: {feat_dims}")
        print(f"  Feature shapes: {[tuple(f.shape) for f in dummy_feats]}")
        
        # Pyramid Pooling Module
        self.ppm = PPM(feat_dims[-1], feat_dims[-1], pool_scales=[1, 2, 3, 6])
        
        # UPerNet decoder with MSCA + LBFR
        self.decoder1 = UPerNetDecoderBlock(feat_dims[3], feat_dims[2], feat_dims[2], 
                                           use_msca=True, use_lbfr=True)
        self.decoder2 = UPerNetDecoderBlock(feat_dims[2], feat_dims[1], feat_dims[1],
                                           use_msca=True, use_lbfr=True)
        self.decoder3 = UPerNetDecoderBlock(feat_dims[1], feat_dims[0], feat_dims[0],
                                           use_msca=True, use_lbfr=True)
        self.decoder4 = UPerNetDecoderBlock(feat_dims[0], 0, 48,
                                           use_msca=True, use_lbfr=False)
        
        # Output head
        self.output_head = nn.Sequential(
            nn.Conv2d(48, 32, 3, padding=1),
            nn.BatchNorm2d(32),
            nn.GELU(),
            nn.Dropout2d(0.05),
            nn.Conv2d(32, num_classes, 1)
        )
    
    def forward(self, x):
        # Input: (B, 3, 256, 256)
        
        # Swin Transformer encoder
        feats = self.backbone(x)
        
        # Swin outputs in [B, H, W, C] format, convert to [B, C, H, W]
        if len(feats[0].shape) == 4 and feats[0].shape[1] < feats[0].shape[-1]:
            feats = [f.permute(0, 3, 1, 2).contiguous() for f in feats]
        
        # Debug: Print feature shapes (only once)
        if not hasattr(self, '_shapes_printed'):
            print(f"\n  Feature shapes from backbone (after permute):")
            for i, feat in enumerate(feats):
                print(f"    Level {i}: {feat.shape}")
            self._shapes_printed = True
        
        # Pyramid Pooling Module
        x = self.ppm(feats[3])
        
        # UPerNet decoder
        x = self.decoder1(x, feats[2])
        x = self.decoder2(x, feats[1])
        x = self.decoder3(x, feats[0])
        x = self.decoder4(x, None)
        
        # Output head
        x = self.output_head(x)
        
        # Upsample to input size (64×64 -> 256×256)
        x = F.interpolate(x, size=(256, 256), mode='bilinear', align_corners=False)
        
        return x

# ============================================================================
# Data Loading (Matching MobileNet)
# ============================================================================

def load_images_masks(images_folder, masks_folder, img_size=(256, 256)):
    """Load images and masks from folders (matching MobileNet)"""
    images, masks = [], []
    image_files = sorted(os.listdir(images_folder))
    mask_files = sorted(os.listdir(masks_folder))
    mask_dict = {os.path.splitext(f)[0]: f for f in mask_files}
    
    for img_file in image_files:
        base = os.path.splitext(img_file)[0]
        if base in mask_dict:
            # Load and preprocess image
            img = cv2.imread(os.path.join(images_folder, img_file))
            img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
            img = cv2.resize(img, img_size)
            
            # Load mask
            mask = cv2.imread(os.path.join(masks_folder, mask_dict[base]), cv2.IMREAD_GRAYSCALE)
            mask = cv2.resize(mask, img_size, interpolation=cv2.INTER_NEAREST)
            
            # Normalize mask values to 0, 1, 2
            unique_vals = np.unique(mask)
            if len(unique_vals) > 3 or np.max(unique_vals) > 2:
                mask = np.clip(mask // 85, 0, 2).astype(np.uint8)
            
            images.append(img)
            masks.append(mask)
    
    return np.array(images, dtype=np.float32), np.array(masks, dtype=np.uint8)

def load_all_data(root_dir):
    """Load all data and merge (matching MobileNet)"""
    all_images = []
    all_masks = []
    
    for split in ["train", "val", "test"]:
        print(f"Loading {split} data...")
        imgs, msks = load_images_masks(
            os.path.join(root_dir, split, "Images"),
            os.path.join(root_dir, split, "Masks")
        )
        all_images.append(imgs)
        all_masks.append(msks)
        print(f"  {split}: {imgs.shape[0]} samples")
    
    all_images = np.concatenate(all_images, axis=0)
    all_masks = np.concatenate(all_masks, axis=0)
    print(f"\nTotal samples: {all_images.shape[0]}")
    print(f"Image shape: {all_images.shape[1:]}")
    print(f"Mask values: {np.unique(all_masks)}")
    
    return all_images, all_masks

def create_new_split(all_images, all_masks):
    """Create new split: 600 train, 200 val, 400 test (matching MobileNet)"""
    print("\nCreating new split (600-200-400)...")
    
    # Split: 600 train, 200 val, 400 test (randomized)
    X_temp, X_test, y_temp, y_test = train_test_split(
        all_images, all_masks, test_size=400, random_state=42, stratify=None
    )
    X_train, X_val, y_train, y_val = train_test_split(
        X_temp, y_temp, test_size=200, random_state=42, stratify=None
    )
    
    print(f"  Train: {X_train.shape[0]} samples")
    print(f"  Val: {X_val.shape[0]} samples")
    print(f"  Test: {X_test.shape[0]} samples")
    
    return (X_train, y_train), (X_val, y_val), (X_test, y_test)

# ============================================================================
# PyTorch Dataset
# ============================================================================

class REFUGEDataset(Dataset):
    """REFUGE dataset with augmentation (matching MobileNet)"""
    def __init__(self, images, masks, augment=None):
        self.images = images
        self.masks = masks
        self.augment = augment
        
        # Transforms
        if augment is not None:
            # Training with augmentation
            self.transform = A.Compose([
                A.RandomBrightnessContrast(brightness_limit=0.15, contrast_limit=0.15, p=0.7),
                A.HueSaturationValue(hue_shift_limit=10, sat_shift_limit=15, val_shift_limit=10, p=0.5),
                A.GaussianBlur(blur_limit=3, p=0.3),
                A.RandomGamma(gamma_limit=(80, 120), p=0.3),
                A.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
                ToTensorV2()
            ])
        else:
            # Validation/Test without augmentation
            self.transform = A.Compose([
                A.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
                ToTensorV2()
            ])
    
    def __len__(self):
        return len(self.images)
    
    def __getitem__(self, idx):
        image = self.images[idx]
        mask = self.masks[idx]
        
        # Apply transforms
        augmented = self.transform(image=image, mask=mask)
        image = augmented['image']
        mask = augmented['mask']
        
        return image, mask.long()

# ============================================================================
# Loss Functions (Matching Keras ODFormer)
# ============================================================================

class FocalDiceLoss(nn.Module):
    """
    Combined Focal + Dice Loss (Optimized Configuration)
    
    Final optimal hyperparameters:
    - alpha=[0.1, 1.4, 1.6]: Background, Disc, Cup weights (asymmetric for stability)
    - gamma=2.5: Strong hard example focusing
    - focal_weight=0.6, dice_weight=0.4: Balanced loss combination
    """
    def __init__(self, alpha=[0.1, 1.4, 1.6], gamma=2.5, focal_weight=0.6, dice_weight=0.4):
        super(FocalDiceLoss, self).__init__()
        self.alpha = torch.tensor(alpha)
        self.gamma = gamma
        self.focal_weight = focal_weight
        self.dice_weight = dice_weight
    
    def focal_loss(self, pred, target):
        """Focal loss with class weights"""
        self.alpha = self.alpha.to(pred.device)
        
        # Get class probabilities
        pred = F.softmax(pred, dim=1)
        pred = torch.clamp(pred, 1e-7, 1.0 - 1e-7)
        
        # One-hot encode targets
        target_one_hot = F.one_hot(target, num_classes=pred.shape[1]).permute(0, 3, 1, 2).float()
        
        # Focal loss per class
        focal = 0.0
        for c in range(pred.shape[1]):
            pred_c = pred[:, c, :, :]
            target_c = target_one_hot[:, c, :, :]
            
            ce = -target_c * torch.log(pred_c)
            focal_c = self.alpha[c] * torch.pow(1 - pred_c, self.gamma) * ce
            focal += focal_c.mean()
        
        return focal
    
    def dice_loss(self, pred, target):
        """Dice loss with class weights"""
        pred = F.softmax(pred, dim=1)
        target_one_hot = F.one_hot(target, num_classes=pred.shape[1]).permute(0, 3, 1, 2).float()
        
        class_weights = torch.tensor([0.1, 1.5, 1.5]).to(pred.device)
        dice = 0.0
        
        for c in range(pred.shape[1]):
            pred_c = pred[:, c, :, :].contiguous().view(-1)
            target_c = target_one_hot[:, c, :, :].contiguous().view(-1)
            
            intersection = (pred_c * target_c).sum()
            union = pred_c.sum() + target_c.sum()
            
            dice_c = 1 - (2.0 * intersection + 1e-7) / (union + 1e-7)
            dice += class_weights[c] * dice_c
        
        return dice / class_weights.sum()
    
    def forward(self, pred, target):
        """Combined loss"""
        focal = self.focal_loss(pred, target)
        dice = self.dice_loss(pred, target)
        return self.focal_weight * focal + self.dice_weight * dice

# ============================================================================
# Metrics (Matching MobileNet)
# ============================================================================

def dice_coefficient(preds, targets, num_classes=3):
    """Calculate Dice coefficient per class"""
    dice_scores = []
    preds = preds.argmax(dim=1)
    
    for c in range(num_classes):
        pred_c = (preds == c).float()
        target_c = (targets == c).float()
        
        intersection = (pred_c * target_c).sum()
        union = pred_c.sum() + target_c.sum()
        
        dice = (2. * intersection + 1e-7) / (union + 1e-7)
        dice_scores.append(dice.item())
    
    return dice_scores

def dice_coefficient_np(y_true, y_pred, num_classes=3):
    """Calculate Dice coefficient per class using numpy (for final evaluation)"""
    dice_scores = []
    
    for c in range(num_classes):
        y_true_c = (y_true == c).astype(np.float32)
        y_pred_c = (y_pred == c).astype(np.float32)
        
        intersection = np.sum(y_true_c * y_pred_c)
        union = np.sum(y_true_c) + np.sum(y_pred_c)
        
        dice = (2. * intersection + 1e-7) / (union + 1e-7)
        dice_scores.append(dice)
    
    return dice_scores

def iou_coefficient_np(y_true, y_pred, num_classes=3):
    """Calculate IoU coefficient per class using numpy (for final evaluation)"""
    iou_scores = []
    
    for c in range(num_classes):
        y_true_c = (y_true == c).astype(np.float32)
        y_pred_c = (y_pred == c).astype(np.float32)
        
        intersection = np.sum(y_true_c * y_pred_c)
        union = np.sum(y_true_c) + np.sum(y_pred_c) - intersection
        
        iou = (intersection + 1e-7) / (union + 1e-7)
        iou_scores.append(iou)
    
    return iou_scores

def iou_coefficient(preds, targets, num_classes=3):
    """Calculate IoU coefficient per class"""
    iou_scores = []
    preds = preds.argmax(dim=1)
    
    for c in range(num_classes):
        pred_c = (preds == c).float()
        target_c = (targets == c).float()
        
        intersection = (pred_c * target_c).sum()
        union = pred_c.sum() + target_c.sum() - intersection
        
        iou = (intersection + 1e-7) / (union + 1e-7)
        iou_scores.append(iou.item())
    
    return iou_scores

# ============================================================================
# Training Loop (Matching MobileNet Configuration)
# ============================================================================

def train_epoch(model, dataloader, criterion, optimizer, device):
    """Train for one epoch"""
    model.train()
    total_loss = 0
    dice_scores = [0, 0, 0]
    iou_scores = [0, 0, 0]
    
    pbar = tqdm(dataloader, total=len(dataloader), bar_format='{l_bar}{bar}| {n_fmt}/{total_fmt} [{elapsed}<{remaining}, {rate_fmt}]')
    
    for batch_idx, (images, masks) in enumerate(pbar):
        images = images.to(device)
        masks = masks.to(device)
        
        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, masks)
        
        loss.backward()
        
        # Tighter gradient clipping for stability
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=0.5)
        
        optimizer.step()
        
        total_loss += loss.item()
        
        # Calculate metrics
        batch_dice = dice_coefficient(outputs.detach(), masks)
        batch_iou = iou_coefficient(outputs.detach(), masks)
        
        for i in range(3):
            dice_scores[i] += batch_dice[i]
            iou_scores[i] += batch_iou[i]
        
        # Update progress bar
        current_loss = total_loss / (batch_idx + 1)
        current_dice = [d / (batch_idx + 1) for d in dice_scores]
        current_iou = [i / (batch_idx + 1) for i in iou_scores]
        
        pbar.set_postfix_str(
            f"loss: {current_loss:.4f} - "
            f"dice_class_0: {current_dice[0]:.4f} - "
            f"dice_class_1: {current_dice[1]:.4f} - "
            f"dice_class_2: {current_dice[2]:.4f} - "
            f"dice_coef_multiclass: {sum(current_dice)/3:.4f} - "
            f"iou_class_0: {current_iou[0]:.4f} - "
            f"iou_class_1: {current_iou[1]:.4f} - "
            f"iou_class_2: {current_iou[2]:.4f} - "
            f"iou_coef_multiclass: {sum(current_iou)/3:.4f}"
        )
    
    pbar.close()
    
    avg_loss = total_loss / len(dataloader)
    avg_dice = [d / len(dataloader) for d in dice_scores]
    avg_iou = [i / len(dataloader) for i in iou_scores]
    
    return avg_loss, avg_dice, avg_iou

def validate(model, dataloader, criterion, device):
    """Validate the model"""
    model.eval()
    total_loss = 0
    dice_scores = [0, 0, 0]
    iou_scores = [0, 0, 0]
    
    with torch.no_grad():
        for images, masks in dataloader:
            images = images.to(device)
            masks = masks.to(device)
            
            outputs = model(images)
            loss = criterion(outputs, masks)
            
            total_loss += loss.item()
            
            batch_dice = dice_coefficient(outputs, masks)
            batch_iou = iou_coefficient(outputs, masks)
            
            for i in range(3):
                dice_scores[i] += batch_dice[i]
                iou_scores[i] += batch_iou[i]
    
    avg_loss = total_loss / len(dataloader)
    avg_dice = [d / len(dataloader) for d in dice_scores]
    avg_iou = [i / len(dataloader) for i in iou_scores]
    
    return avg_loss, avg_dice, avg_iou

# ============================================================================
# Evaluation Functions (Matching MobileNet)
# ============================================================================

def evaluate_model(model, test_loader, device):
    """Comprehensive model evaluation (matching MobileNet style)"""
    print("\nEvaluating model on test set...")
    
    model.eval()
    all_preds = []
    all_targets = []
    
    with torch.no_grad():
        for images, masks in test_loader:
            images = images.to(device)
            outputs = model(images)
            preds = outputs.argmax(dim=1).cpu().numpy()
            
            all_preds.extend(preds.flatten())
            all_targets.extend(masks.numpy().flatten())
    
    all_preds = np.array(all_preds)
    all_targets = np.array(all_targets)
    
    # Overall metrics
    precision = precision_score(all_targets, all_preds, average='weighted', zero_division=0)
    recall = recall_score(all_targets, all_preds, average='weighted', zero_division=0)
    f1 = f1_score(all_targets, all_preds, average='weighted', zero_division=0)
    cm = confusion_matrix(all_targets, all_preds)
    
    print(f"\nTest Metrics:")
    print(f"  Precision (weighted): {precision:.4f}")
    print(f"  Recall (weighted): {recall:.4f}")
    print(f"  F-score (weighted): {f1:.4f}")
    print(f"\nConfusion Matrix:")
    print(cm)
    
    # Per-class Dice and IoU using numpy (matching MobileNet)
    all_preds_2d = all_preds.reshape(-1, 256, 256)
    all_targets_2d = all_targets.reshape(-1, 256, 256)
    
    dice_scores = dice_coefficient_np(all_targets_2d, all_preds_2d)
    iou_scores = iou_coefficient_np(all_targets_2d, all_preds_2d)
    
    print(f"\nPer-class Dice on TEST set:")
    class_names = ["Background", "Disc", "Cup"]
    for i, cname in enumerate(class_names):
        print(f"  {cname} - Dice: {dice_scores[i]:.4f}, IoU: {iou_scores[i]:.4f}")
    
    # Overall multiclass Dice and IoU
    overall_dice = np.mean(dice_scores)
    overall_iou = np.mean(iou_scores)
    
    print(f"\nOverall multiclass Dice: {overall_dice:.4f}")
    print(f"Overall multiclass IoU: {overall_iou:.4f}")
    
    return precision, recall, f1, dice_scores, iou_scores, overall_dice, overall_iou

# ============================================================================
# Main Training Function
# ============================================================================

def main():
    """Main training script (matching MobileNet configuration)"""
    
    # Configuration (MATCHING MOBILENET)
    BATCH_SIZE = 8
    NUM_EPOCHS = 100
    LEARNING_RATE = 1e-3  # Same as MobileNet
    WEIGHT_DECAY = 1e-4
    NUM_WORKERS = 2
    DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    
    # Paths
    DATA_ROOT = "/kaggle/input/refuge/REFUGE/"
    
    print("="*80)
    print("ODFormer Training (MobileNet Comparison Configuration)")
    print("="*80)
    print(f"\nConfiguration:")
    print(f"  Device: {DEVICE}")
    print(f"  Batch Size: {BATCH_SIZE}")
    print(f"  Epochs: {NUM_EPOCHS}")
    print(f"  Learning Rate: {LEARNING_RATE * 0.1} (1e-4, low for stability)")
    print(f"  Loss: 0.6×Focal(γ=2.5,α=[0.1,1.4,1.6]) + 0.4×Dice (OPTIMIZED)")
    print(f"  Gradient Clipping: 0.5 (tight)")
    print(f"  Warmup: 3 epochs")
    print(f"  Optimizer: AdamW (better for transformers)")
    print(f"  LR Scheduler: ReduceLROnPlateau (factor=0.5, patience=5)")
    print(f"  Early Stopping: patience=10 (matching MobileNet: patience=8)")
    print(f"  Input Size: 256×256 (same as MobileNet)")
    print(f"  Split: 600 train, 200 val, 400 test (same as MobileNet)")
    print(f"  Augmentation: Same as MobileNet")
    print(f"  Expected Performance: ~88.96% Dice (Disc+Cup), ~80.18% IoU")
    
    # Load and split data (matching MobileNet)
    print(f"\nLoading REFUGE dataset...")
    all_images, all_masks = load_all_data(DATA_ROOT)
    (X_train, y_train), (X_val, y_val), (X_test, y_test) = create_new_split(all_images, all_masks)
    
    # Normalize images to [0, 1]
    X_train = X_train / 255.0
    X_val = X_val / 255.0
    X_test = X_test / 255.0
    
    # Create datasets
    train_dataset = REFUGEDataset(X_train, y_train, augment=True)
    val_dataset = REFUGEDataset(X_val, y_val, augment=None)
    test_dataset = REFUGEDataset(X_test, y_test, augment=None)
    
    train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True, 
                              num_workers=NUM_WORKERS, pin_memory=True)
    val_loader = DataLoader(val_dataset, batch_size=BATCH_SIZE, shuffle=False,
                           num_workers=NUM_WORKERS, pin_memory=True)
    test_loader = DataLoader(test_dataset, batch_size=BATCH_SIZE, shuffle=False,
                            num_workers=NUM_WORKERS, pin_memory=True)
    
    # Create model
    print(f"\nBuilding ODFormer model...")
    model = ODFormer(num_classes=3, pretrained=True, backbone='swin_tiny_patch4_window7_224')
    model = model.to(DEVICE)
    
    # Count parameters
    total_params = sum(p.numel() for p in model.parameters())
    trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
    print(f"  Total parameters: {total_params:,}")
    print(f"  Trainable parameters: {trainable_params:,}")
    
    # Loss and optimizer (matching Keras ODFormer.py)
    # Use Focal + Dice loss like the working Keras implementation
    criterion = FocalDiceLoss(alpha=[0.1, 1.4, 1.6], gamma=2.5, focal_weight=0.6, dice_weight=0.4)
    
    # Use AdamW (better for transformers) with low learning rate
    optimizer = torch.optim.AdamW(model.parameters(), lr=LEARNING_RATE * 0.1, weight_decay=WEIGHT_DECAY)
    
    # Learning rate scheduler (matching MobileNet)
    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
        optimizer, mode='min', factor=0.5, patience=5, verbose=True, min_lr=1e-7
    )
    
    # Training loop
    print(f"\n{'='*80}")
    print(f"Starting Training")
    print(f"{'='*80}\n")
    
    best_val_loss = float('inf')
    best_dice = 0.0
    patience_counter = 0
    patience = 10  # Early stopping patience (more tolerant)
    
    history = {
        'train_loss': [], 'val_loss': [],
        'train_dice': [], 'val_dice': [],
        'train_iou': [], 'val_iou': []
    }
    
    for epoch in range(NUM_EPOCHS):
        # Print epoch header
        print(f"Epoch {epoch+1}/{NUM_EPOCHS}")
        
        # Learning rate warmup for first 3 epochs
        if epoch < 3:
            warmup_factor = (epoch + 1) / 3.0
            for param_group in optimizer.param_groups:
                param_group['lr'] = LEARNING_RATE * 0.1 * warmup_factor
        
        # Train
        train_loss, train_dice, train_iou = train_epoch(model, train_loader, criterion, optimizer, DEVICE)
        
        # Check if improved before validation
        improved_msg = ""
        if val_loss < best_val_loss if epoch > 0 else True:
            pass  # Will be set after validation
        
        # Validate (no progress bar, silent)
        val_loss, val_dice, val_iou = validate(model, val_loader, criterion, DEVICE)
        
        # Update learning rate
        old_lr = optimizer.param_groups[0]['lr']
        scheduler.step(val_loss)
        new_lr = optimizer.param_groups[0]['lr']
        
        # Store history
        history['train_loss'].append(train_loss)
        history['val_loss'].append(val_loss)
        history['train_dice'].append(train_dice)
        history['val_dice'].append(val_dice)
        history['train_iou'].append(train_iou)
        history['val_iou'].append(val_iou)
        
        # Save best model and set improved message
        if val_loss < best_val_loss:
            improved_msg = f": val_loss improved from {best_val_loss:.5f} to {val_loss:.5f}, saving model to odformer_mobilenet_comparison_best.pth"
            best_val_loss = val_loss
            best_dice = val_dice[1]  # Disc dice
            torch.save({
                'epoch': epoch,
                'model_state_dict': model.state_dict(),
                'optimizer_state_dict': optimizer.state_dict(),
                'val_loss': val_loss,
                'val_dice': val_dice,
            }, 'odformer_mobilenet_comparison_best.pth')
            patience_counter = 0
        else:
            improved_msg = f": val_loss did not improve from {best_val_loss:.5f}"
            patience_counter += 1
        
        # Print single summary line (matching Keras format exactly)
        print(f"Epoch {epoch+1}{improved_msg}")
        print(f"{len(train_loader)}/{len(train_loader)} ━━━━━━━━━━━━━━━━━━━━ "
              f"loss: {train_loss:.4f} - "
              f"dice_class_0: {train_dice[0]:.4f} - "
              f"dice_class_1: {train_dice[1]:.4f} - "
              f"dice_class_2: {train_dice[2]:.4f} - "
              f"dice_coef_multiclass: {sum(train_dice)/3:.4f} - "
              f"iou_class_0: {train_iou[0]:.4f} - "
              f"iou_class_1: {train_iou[1]:.4f} - "
              f"iou_class_2: {train_iou[2]:.4f} - "
              f"iou_coef_multiclass: {sum(train_iou)/3:.4f} - "
              f"val_loss: {val_loss:.4f} - "
              f"val_dice_class_0: {val_dice[0]:.4f} - "
              f"val_dice_class_1: {val_dice[1]:.4f} - "
              f"val_dice_class_2: {val_dice[2]:.4f} - "
              f"val_dice_coef_multiclass: {sum(val_dice)/3:.4f} - "
              f"val_iou_class_0: {val_iou[0]:.4f} - "
              f"val_iou_class_1: {val_iou[1]:.4f} - "
              f"val_iou_class_2: {val_iou[2]:.4f} - "
              f"val_iou_coef_multiclass: {sum(val_iou)/3:.4f} - "
              f"learning_rate: {new_lr:.4e}")
        
        # Early stopping
        if patience_counter >= patience:
            print(f"\nEarly stopping triggered after {epoch+1} epochs")
            break
    
    # Load best model and evaluate
    print(f"\n{'='*80}")
    print(f"Final Evaluation on Test Set")
    print(f"{'='*80}\n")
    
    print(f"Restoring model weights from the end of the best epoch: {checkpoint['epoch']+1 if 'checkpoint' in locals() else 'N/A'}")
    checkpoint = torch.load('odformer_mobilenet_comparison_best.pth')
    print(f"Restoring model weights from the end of the best epoch: {checkpoint['epoch']+1}")
    model.load_state_dict(checkpoint['model_state_dict'])
    
    test_loss, test_dice, test_iou = validate(model, test_loader, criterion, DEVICE)
    
    print(f"\nTest Results:")
    print(f"  Test Loss: {test_loss:.4f}")
    print(f"  Test Dice - BG: {test_dice[0]:.4f}, Disc: {test_dice[1]:.4f}, Cup: {test_dice[2]:.4f}")
    print(f"  Test IoU  - BG: {test_iou[0]:.4f}, Disc: {test_iou[1]:.4f}, Cup: {test_iou[2]:.4f}")
    print(f"  Average Dice (Disc+Cup): {(test_dice[1] + test_dice[2]) / 2:.4f}")
    print(f"  Average IoU (Disc+Cup): {(test_iou[1] + test_iou[2]) / 2:.4f}")
    
    # Detailed evaluation (matching MobileNet style)
    precision, recall, f1, dice_scores_detailed, iou_scores_detailed, overall_dice, overall_iou = evaluate_model(model, test_loader, DEVICE)
    
    # Plot training history
    plot_training_history(history)
    
    # Visualize predictions (matching MobileNet style)
    print(f"\n{'='*60}")
    print(f"SAMPLE PREDICTIONS")
    print(f"{'='*60}")
    visualize_predictions(model, test_loader, DEVICE, "ODFormer", num_samples=3)
    
    # Final Results Summary (matching MobileNet style)
    print(f"\n{'='*100}")
    print(f"FINAL RESULTS SUMMARY")
    print(f"{'='*100}")
    
    print(f"\nModel Performance:")
    print(f"  Overall Dice: {overall_dice:.4f}")
    print(f"  Overall IoU: {overall_iou:.4f}")
    print(f"  Parameters: {total_params:,}")
    
    print(f"\nDetailed Test Metrics:")
    print(f"  Test Loss: {test_loss:.4f}")
    print(f"  Precision (weighted): {precision:.4f}")
    print(f"  Recall (weighted): {recall:.4f}")
    print(f"  F-score (weighted): {f1:.4f}")
    
    print(f"\nPer-class Results:")
    class_names = ["Background", "Disc", "Cup"]
    for i, cname in enumerate(class_names):
        print(f"  {cname} - Dice: {dice_scores_detailed[i]:.4f}, IoU: {iou_scores_detailed[i]:.4f}")
    
    print(f"\n{'='*100}")
    print(f"Training Completed Successfully!")
    print(f"{'='*100}\n")
    print(f"Best model saved as: odformer_mobilenet_comparison_best.pth")

# ============================================================================
# Plotting Functions
# ============================================================================

def visualize_predictions(model, test_loader, device, model_name="ODFormer", num_samples=3):
    """Visualize sample predictions (matching MobileNet style) - Display only, no saving"""
    
    print(f"\nGenerating sample predictions for {model_name}...")
    
    model.eval()
    images_list = []
    masks_list = []
    preds_list = []
    
    with torch.no_grad():
        for images, masks in test_loader:
            if len(images_list) >= num_samples:
                break
            
            images = images.to(device)
            outputs = model(images)
            preds = outputs.argmax(dim=1).cpu().numpy()
            
            # Move to CPU and denormalize images (matching ODFormer.ipynb)
            images_np = images.cpu().numpy()
            
            for i in range(images.shape[0]):
                if len(images_list) >= num_samples:
                    break
                
                # Denormalize image (reverse ImageNet normalization)
                # Images are in [C, H, W] format, need to transpose to [H, W, C]
                img = images_np[i].transpose(1, 2, 0)  # [3, 256, 256] -> [256, 256, 3]
                mean = np.array([0.485, 0.456, 0.406])
                std = np.array([0.229, 0.224, 0.225])
                img = img * std + mean  # Denormalize
                img = np.clip(img, 0, 1)  # Ensure values are in [0, 1]
                
                images_list.append(img)
                masks_list.append(masks[i].numpy())
                preds_list.append(preds[i])
    
    # Visualize - Display only
    for i in range(num_samples):
        plt.figure(figsize=(16, 4))
        
        plt.subplot(1, 4, 1)
        plt.imshow(images_list[i])
        plt.title("Original Image")
        plt.axis('off')
        
        plt.subplot(1, 4, 2)
        plt.imshow(masks_list[i], cmap='jet', vmin=0, vmax=2)
        plt.title("Ground Truth")
        plt.axis('off')
        
        plt.subplot(1, 4, 3)
        plt.imshow(preds_list[i], cmap='jet', vmin=0, vmax=2)
        plt.title("Prediction")
        plt.axis('off')
        
        plt.subplot(1, 4, 4)
        overlay = images_list[i].copy()
        pred_mask = preds_list[i]
        overlay[pred_mask == 1] = [1, 0, 0]  # Red for disc
        overlay[pred_mask == 2] = [0, 1, 0]  # Green for cup
        plt.imshow(overlay)
        plt.title("Overlay")
        plt.axis('off')
        
        plt.suptitle(f"{model_name} - Sample {i+1}")
        plt.tight_layout()
        plt.show()
        plt.close()
    
    print(f"✓ {num_samples} prediction samples displayed")

def plot_training_history(history):
    """Plot training curves (matching MobileNet style) - Display only, no saving"""
    epochs = range(1, len(history['train_loss']) + 1)
    
    plt.figure(figsize=(20, 5))
    
    # Loss plot
    plt.subplot(1, 4, 1)
    plt.plot(epochs, history['train_loss'], 'b-', label='Train Loss')
    plt.plot(epochs, history['val_loss'], 'r-', label='Val Loss')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.title('Training and Validation Loss')
    plt.legend()
    plt.grid(True)
    
    # Dice - Background
    plt.subplot(1, 4, 2)
    train_dice_bg = [d[0] for d in history['train_dice']]
    val_dice_bg = [d[0] for d in history['val_dice']]
    plt.plot(epochs, train_dice_bg, 'b-', label='Train')
    plt.plot(epochs, val_dice_bg, 'r-', label='Val')
    plt.xlabel('Epoch')
    plt.ylabel('Dice Coefficient')
    plt.title('Dice - Background')
    plt.legend()
    plt.grid(True)
    
    # Dice - Disc
    plt.subplot(1, 4, 3)
    train_dice_disc = [d[1] for d in history['train_dice']]
    val_dice_disc = [d[1] for d in history['val_dice']]
    plt.plot(epochs, train_dice_disc, 'b-', label='Train')
    plt.plot(epochs, val_dice_disc, 'r-', label='Val')
    plt.xlabel('Epoch')
    plt.ylabel('Dice Coefficient')
    plt.title('Dice - Disc')
    plt.legend()
    plt.grid(True)
    
    # Dice - Cup
    plt.subplot(1, 4, 4)
    train_dice_cup = [d[2] for d in history['train_dice']]
    val_dice_cup = [d[2] for d in history['val_dice']]
    plt.plot(epochs, train_dice_cup, 'b-', label='Train')
    plt.plot(epochs, val_dice_cup, 'r-', label='Val')
    plt.xlabel('Epoch')
    plt.ylabel('Dice Coefficient')
    plt.title('Dice - Cup')
    plt.legend()
    plt.grid(True)
    
    plt.tight_layout()
    plt.show()
    print(f"\n✓ Training history plots displayed")
    plt.close()
    
    # Additional plots - Dice Coefficient multiclass and Accuracy over epochs
    plt.figure(figsize=(15, 5))
    
    # Multiclass Dice
    plt.subplot(1, 3, 1)
    train_dice_multiclass = [(d[0] + d[1] + d[2]) / 3 for d in history['train_dice']]
    val_dice_multiclass = [(d[0] + d[1] + d[2]) / 3 for d in history['val_dice']]
    plt.plot(epochs, train_dice_multiclass, 'b-', label='Train')
    plt.plot(epochs, val_dice_multiclass, 'r-', label='Val')
    plt.xlabel('Epoch')
    plt.ylabel('Dice Coefficient')
    plt.title('Dice Coefficient Multiclass over Epochs')
    plt.legend()
    plt.grid(True)
    
    # Multiclass IoU
    plt.subplot(1, 3, 2)
    train_iou_multiclass = [(i[0] + i[1] + i[2]) / 3 for i in history['train_iou']]
    val_iou_multiclass = [(i[0] + i[1] + i[2]) / 3 for i in history['val_iou']]
    plt.plot(epochs, train_iou_multiclass, 'b-', label='Train')
    plt.plot(epochs, val_iou_multiclass, 'r-', label='Val')
    plt.xlabel('Epoch')
    plt.ylabel('IoU Coefficient')
    plt.title('IoU Coefficient Multiclass over Epochs')
    plt.legend()
    plt.grid(True)
    
    # Learning Rate
    plt.subplot(1, 3, 3)
    # Extract LR from training (would need to track in history)
    plt.title('Learning Rate Schedule')
    plt.xlabel('Epoch')
    plt.ylabel('Learning Rate')
    plt.grid(True)
    plt.text(0.5, 0.5, 'LR: 1e-4 with ReduceLROnPlateau\n(factor=0.5, patience=5)', 
             ha='center', va='center', transform=plt.gca().transAxes)
    
    plt.tight_layout()
    plt.show()
    print(f"✓ Additional metrics plots displayed")
    plt.close()

# ============================================================================
# Entry Point
# ============================================================================

if __name__ == '__main__':
    main()
```
